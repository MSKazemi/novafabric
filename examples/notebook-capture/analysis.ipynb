# Notebook capture — a worked example

A deterministic, standard-library-only analysis. No API key, no network, no
third-party imports, so it runs identically on any machine.

Run it under capture with the command in `README.md`; the whole notebook
execution becomes one Run Capsule.

```python
import json
import os
import statistics
from pathlib import Path

# A fixed dataset, so two runs of this notebook are genuinely comparable.
LATENCIES_MS = [
    118, 122, 119, 131, 127, 124, 402, 121, 117, 129,
    125, 120, 133, 126, 118, 141, 123, 119, 388, 122,
]
print(f"samples: {len(LATENCIES_MS)}")
```

```python
summary = {
    "count": len(LATENCIES_MS),
    "mean_ms": round(statistics.mean(LATENCIES_MS), 2),
    "median_ms": statistics.median(LATENCIES_MS),
    "stdev_ms": round(statistics.stdev(LATENCIES_MS), 2),
    "min_ms": min(LATENCIES_MS),
    "max_ms": max(LATENCIES_MS),
}

# The tail is what a latency SLO is actually judged on, so report it explicitly
# rather than letting the mean hide the two 380ms+ outliers.
ordered = sorted(LATENCIES_MS)
summary["p95_ms"] = ordered[max(0, round(0.95 * len(ordered)) - 1)]

for key, value in summary.items():
    print(f"{key:12s} {value}")
```

```python
# Write the result where the capture can pick it up as an output artifact.
out_dir = Path(os.environ.get("NOTEBOOK_OUTPUT_DIR", "."))
out_dir.mkdir(parents=True, exist_ok=True)
result_path = out_dir / "results.json"
result_path.write_text(json.dumps(summary, indent=2, sort_keys=True) + "\n")

print(f"notebook: wrote {result_path}")
```
